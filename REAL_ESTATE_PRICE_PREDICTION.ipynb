```python
# SO NOW HERE IF WE HAVE RANGE WE TAKE --> AVERAGE OF THOSE
# IF WE HAVE VALUES WITH STRING WE CAN IGNORE THOSE VALUES
# IF WE NEED A STILL SOPISTSTICAED MANNER WE CAN DO UNIT CONVERSION 

def convert_sqft_to_num(x):
    tokens = x.split("-")
    if len(tokens) == 2:
        return (float(tokens[0])+float(tokens[1]))/2
    try:
         return float(x)
    except:
         return None
    
```

```python
# AS WE HAVE ADDED NEW COLUMN LET'S RUN SOME BASIC STATISTICS ON THE COLUMN ADDED
# WE CAN CLEARLY OBSERVE BY LOOKING AT MIN AND MAX VALUES WHERE VALUE 371 PER SQFT IS TOO LOW
# AND THE MAX VALUE 76530 PER SQFT IS TOO HIGH ,IT MAY BE IN ONLY RARE CASES 
# NOW LET'S REMOVE PRICE PER SQUARE FEET PER LOCATION AS SOME LOCATION WILL HAVE HIGHER PRICE 
# SOME LOCATIONS WILL HAVE LESS PRICE
# FOR DATAPOINTS WE HAVE TO FIND MEAN AND STANDARD DEVIATION 
# FILTER OUT THE DATAPOINTS WHICH ARE BEYOND ONE SD
# Z_SCORE METHOD

def remove_pps_outliers(df):
    df_out = pd.DataFrame()
    for key, subdf in df.groupby('location'):
        mean = np.mean (subdf.price_per_sqft)
        sd = np.std(subdf.price_per_sqft )
        reduced_df = subdf[(subdf.price_per_sqft>(mean-sd)) & (subdf.price_per_sqft<=(mean+sd))]
        df_out = pd.concat([df_out, reduced_df],ignore_index=True)
    return df_out


outlier_z_score = remove_pps_outliers(outliers)
outlier_z_score.shape
```

## *SCATTERPLOT*

```python
# PROPERTY PRICE 
# FOR 3 BEDROOM APARTMENTS ARE HAVING MORE PRICE THAN 2 BEDROOM APARTMENTS FOR THE SAME SQUAREFOOT AREA
# IN SUCH CASES WE CAN USE SCATTERPLOT TO GET THE VISUALIZATION FOR OUTLIER DETECTION
# AS LOCATION PLAYS A MAJOR ROLE WE CAN USE GROUPBY LOCATION AND use scatterplot FUNCTION

def plot_scatter_chart(df,location) :
    bhk2 = df[(df.location==location) & (df.bedrooms==2)]
    bhk3 = df[(df.location==location) & (df.bedrooms==3)]
    matplotlib.rcParams['figure.figsize'] = (15,10)
    plt.scatter(bhk2.total_sqft,bhk2.price,color='blue',label='2 BHK', s=50)
    plt.scatter(bhk3.total_sqft,bhk3.price,marker='+', color='green' ,label='3 BHK', s=50)
    plt.xlabel("Total Square Feet Area")
    plt.ylabel("Price Per Square Feet")
    plt.title(location)
    plt.legend()


plot_scatter_chart(outlier_z_score, "Rajaji Nagar")

# BLUE POINTS ARE 2 BEDROOM APARTMENTS GREEN POINTS ARE 3 BEDROOM
# WE CAN OBSERVE FOR THE SAME SQUAREFOOT AREA 2 BEDROOM AP PRICE IS MORE THAN 3 BEDROOMS
# WE CAN ROMOVE SUCH KIND OR VALUES
```

```python
plot_scatter_chart(outlier_z_score, "Hebbal")
```

```python
def remove_bhk_outliers(df) :
    exclude_indices = np.array([])
    for location, location_df in df.groupby('location'):
        bhk_stats = {}
        for bhk, bhk_df in location_df.groupby('bedrooms'):
            bhk_stats[bhk] = {
                'mean' : np.mean(bhk_df.price_per_sqft),
                'std': np.std(bhk_df.price_per_sqft),
                'count': bhk_df. shape [0]
           }
        for bhk, bhk_df in location_df.groupby('bedrooms'):
            stats = bhk_stats.get(bhk-1)
            if stats and stats [ 'count']>5:
                exclude_indices = np. append(exclude_indices, bhk_df[bhk_df. price_per_sqft<(stats[ 'mean'])]. index. values)
    return df.drop(exclude_indices, axis='index')


outliers_scatter_plot = remove_bhk_outliers(outlier_z_score)
outliers_scatter_plot.shape
```

```python
# WE CAN DEFINE A FUNCTION FOR PREDICTING THE MODEL WITH ALL THE INPUT FEATURES

def predict_price(location,total_sqft,bath,bedrooms):
    loc_index = np.where(X_train.columns == location)[0][0]

    x = np.zeros(len(X_train.columns))
    x[0] = total_sqft
    x[1] = bath
    x[2] = bedrooms
    if loc_index >= 0:
        x[loc_index] = 1

    return lr.predict([x])[0]
```

## *GRID SEARCH CV()*
   - **WE CAN USE GRID SEARCH CV TO CHECK OUT HOW DIFFERENT MODELS PERFORM ON THE DATASET**

```python
from sklearn.model_selection import GridSearchCV
import warnings
%matplotlib inline
warnings.filterwarnings('ignore')

from sklearn.linear_model import Lasso
from sklearn.tree import DecisionTreeRegressor

# WE CAN DEFINE A FUNCTION WHICH HELPS US TO FIT ALL MODELS TOGETHER

def find_best_model_using_gridsearchcv(X,y) :
    algos = {
        'linear_regression' : {
            'model': LinearRegression(),
            'params': {
                'fit_intercept' : [True],
                'copy_X' : [True],
                'n_jobs' : [None],
                'positive': [False]
            }
        },
        'lasso': {
            'model': Lasso(),
            'params': {
                'alpha': [1,2],
                'selection': ['random', 'cyclic']
            }
        },
        'decision_tree': {
            'model': DecisionTreeRegressor(),
            'params': {
                'criterion' : ['mse', 'friedman_mse'],
                'splitter': ['best','random' ]
            }
        }
    }
    scores = []
    cv = KFold(n_splits = 5,shuffle = True)
    for algo_name ,config in algos.items():
        gs = GridSearchCV(config['model'],config['params'],cv=cv,return_train_score = False)
        gs.fit(X,y)
        scores.append({
            'model': algo_name,
            'best_score' : gs.best_score_,
            'best_params':gs.best_params_
        })
    return pd.DataFrame(scores,columns = ['model','best_score','best_params'])

find_best_model_using_gridsearchcv(X_train,y_train)
        
```
